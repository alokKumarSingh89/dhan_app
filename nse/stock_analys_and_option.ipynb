```python
import pandas as pd
from datetime import datetime, timedelta
from py_vollib.black_scholes_merton.implied_volatility import implied_volatility
from py_vollib.black_scholes import black_scholes
from time import strptime
from database.db import db
import os
```

```python
def process_option_data(flag):
    r = 0.1
    def getDate(data):
        tmp = data.split("-")
        tmp[1] = strptime(tmp[1],'%b').tm_mon
        date_str = datetime(int(tmp[2]),tmp[1],int(tmp[0]),15,30,0)
        return date_str
    def create_sigma(df, flag):
        def calulate_sigma(data):
            price = float(data["LTP"])
            S = float(data["Underlying Value"])
            current_date = getDate(data["Date"])
            expiry_date = getDate(data["Expiry"])
            K = data["Strike Price"]
            t = ((expiry_date - current_date)/timedelta(days=1))/365
            try:
                sigma = implied_volatility(price, S, K, t, r, 0, flag)
                return sigma
            except Exception as e:
                print(e)
                return 0
        
        df['sigma'] = df.apply(calulate_sigma, axis=1)
    for filename in os.listdir(f"./data/{flag}"):
        print(f"processing for {filename}")
        isFileProcess = db["expiry"].find_one({"file":filename})
        if isFileProcess is not None:
            print(f"{filename} alread_process")
            continue
        df = pd.read_csv(f"./data/{flag}/{filename}")
        df.rename(columns=lambda x: x.strip(), inplace=True)
        df = df.drop(['Open','High','Low','Symbol','Turnover * in  ₹ Lakhs',"Premium Turnover ** in   ₹ Lakhs","No. of contracts"], axis=1)
        df = df[df["LTP"] != '-']
        
        create_sigma(df,flag) 
        records = df.to_dict('records')
        expirty_c = df["Expiry"].values[0]
        db["expiry"].insert_one({"expiry_date": expirty_c, "file":filename})
        db[f"historical-{expirty_c[3:]}"].insert_many(records)
        
```

```python
process_option_data('p')
```
