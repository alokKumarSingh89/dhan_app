## NSE ETF Data

```python
import pandas as pd
from database.db import db
```

#### Filter the etf that have volumn more than 10k

```python
df = pd.read_csv("./d.csv")
df.rename(columns=lambda x: x.strip(), inplace=True)
df["VOLUME"] = df["VOLUME"].apply(lambda x : int(x.replace(',','')))
df = df[df["VOLUME"]>10000]
df
```

#### Create Symbol for fyer

```python
new_list = df["SYMBOL"].to_list()
fyers_etf = [{"name": f"NSE:{item}-EQ", "broker":"fyers"} for item in new_list]
dhan_etf = [{"name":f"{item}-ETF","broker":"dhan"} for item in new_list]
fyers_etf
```

```python
db["etf_script"].drop()
db["etf_script"].insert_many(fyers_etf)
db["etf_script"].insert_many(dhan_etf)
```
