```python
import pandas as pd
```

```python
def proccess(data):
    if data["tranaction_type"] == "Sell":
        data["pnl"] = data["entry_price"] - data["exit_price"]
    else:
        data["pnl"] = -data["entry_price"] + data["exit_price"]
    return data
```

```python
df = pd.read_csv(f"test.csv")
df = df.apply(proccess,axis=1)
df["per_loss"] = (df["pnl"]/df["entry_price"])*100
sum = df["per_loss"].sum(axis=0)
sum
# df.tail(20)
```
