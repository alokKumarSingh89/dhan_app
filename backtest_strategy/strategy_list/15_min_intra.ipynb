```python
import pandas as pd
from database.db import db
from datetime import datetime, timedelta
```

```python
def proccess(data):
    if data["transaction"] == "Sell":
        data["pnl"] = data["entry"] - data["exit"]
    else:
        data["pnl"] = -data["entry"] + data["exit"]
    return data
```

```python
def run():
    dict_data = {"script":[],"date":[],"pnl":[]}
    for sy in ["SBIN","TCS","ICICIBANK","INFY","BAJFINANCE"]:
        for year in range(2020,2024):
            df = pd.read_csv(f"backtest/15_min_b_in/{year}-{sy}.csv")
            df = df[df['date']<df['exit_time']]
            df = df.apply(proccess,axis=1)
            df2=df.dropna(axis=0)
            sum = df2["pnl"].sum(axis=0)
            dict_data["date"].append(year)
            dict_data["script"].append(sy)
            dict_data["pnl"].append((sum/df2["entry"].iloc[0])*100)
    return pd.DataFrame(dict_data)
```

```python
df = run()
df.sort_values(by='pnl',ascending=False)
```
